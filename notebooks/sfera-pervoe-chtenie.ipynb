```python
# Концепт «Живое пространство»: печать целиком
import base64, json, urllib.request

LOCAL = 'docs/concept/live-space.md'      # если ноутбук клонировал ветку с концептом
BRANCH = 'arena/cdd63c83-potok'           # ветка концепта (PR #13)

text = None
if os.path.exists(LOCAL):
    text = open(LOCAL, encoding='utf-8').read()
    print('Читаю из локального клона:', LOCAL)
else:
    url = f'https://api.github.com/repos/COL-O/POTOK/contents/docs/concept/live-space.md?ref={BRANCH}'
    with urllib.request.urlopen(url, timeout=30) as r:
        text = base64.b64decode(json.load(r)['content']).decode('utf-8')
    print('В клоне файла нет — читаю через GitHub API из ветки', BRANCH)

print()
print(text)
print()
print('Разделов (##):', text.count('\n## ') + (1 if text.startswith('## ') else 0))
```

## Прогон POTOK на Colab (полигон)

Colab — не «записная книжка», а полигон: ~12 ГБ ОЗУ и CPU (T4 для Go-тестов не нужен и не поможет).

Что делают следующие ячейки:
1. ставят Go 1.21 из официального тарбола (apt в Colab слишком старый) и gcc для `-race`;
2. клонируют **отдельную** копию ветки в `/content/POTOK_RUN` — клон ячейки 2 не трогается;
3. гоняют тесты: корневой модуль, затем `flat/` с `-race`, затем Zero-GC проверки и сборку;
4. результат пишут в `/content/potok-run.log`; последняя ячейка копирует лог на Диск (если он смонтирован).

⚠️ По умолчанию BRANCH = `arena/3bbbb51d-potok` (там живёт полный организм, flat/ и modules/bus/); в нашей ветке `arena/cdd63c83-potok` — только скелет на 71 файл, тестов почти нет.
⚠️ Ошибки прогона — это диагностика, а не «ноутбук сломался».


```python
# 1. Go 1.21 + gcc (для -race). apt-версию не используем.
import os, subprocess
GO_VERSION = '1.21.13'
tgz = f'go{GO_VERSION}.linux-amd64.tar.gz'
url = f'https://go.dev/dl/{tgz}'
if subprocess.run('command -v /usr/local/go/bin/go', shell=True).returncode != 0:
    subprocess.run(['bash','-lc', f'wget -q {url} -O /tmp/{tgz} && tar -C /usr/local -xzf /tmp/{tgz} && rm /tmp/{tgz}'], check=True)
    subprocess.run('ln -sf /usr/local/go/bin/go /usr/local/bin/go && ln -sf /usr/local/go/bin/gofmt /usr/local/bin/gofmt', shell=True, check=True)
if subprocess.run('command -v gcc', shell=True).returncode != 0:
    subprocess.run('apt-get -qq update && apt-get -qq install -y gcc > /dev/null', shell=True, check=True)
os.environ['PATH'] = '/usr/local/go/bin:/root/go/bin:' + os.environ['PATH']
print(subprocess.run(['/usr/local/go/bin/go','version'],capture_output=True,text=True).stdout)
print(subprocess.run(['gcc','--version'],capture_output=True,text=True).stdout.splitlines()[0])
```

```python
# 2. Свежий клон в отдельную папку (клон ячейки 2 не трогаем)
BRANCH = 'arena/3bbbb51d-potok'   # полный организм; для скелета нашей ветки: 'arena/cdd63c83-potok'
RUN_DIR = '/content/POTOK_RUN'
import shutil, os, subprocess
if os.path.exists(RUN_DIR):
    shutil.rmtree(RUN_DIR)
subprocess.run(['git','clone','--depth','1','-b',BRANCH,'https://github.com/COL-O/POTOK.git',RUN_DIR], check=True)
print(subprocess.run(['git','-C',RUN_DIR,'log','--oneline','-1'],capture_output=True,text=True).stdout)
print('плоский модуль go.mod:', open(RUN_DIR + '/flat/go.mod').read().split(chr(10))[:3] if os.path.exists(RUN_DIR + '/flat/go.mod') else 'НЕТ')
```

```python
# 3. Прогон: тесты, Zero-GC, сборка. Всё пишется в /content/potok-run.log
import os, subprocess, time, pathlib
RUN_DIR = '/content/POTOK_RUN'
LOG = pathlib.Path('/content/potok-run.log')
env = dict(os.environ, PATH='/usr/local/go/bin:/root/go/bin:' + os.environ.get('PATH',''))
results = []

def run(title, cmd, cwd=None, timeout=1200, env=env):
    t0 = time.time()
    try:
        r = subprocess.run(cmd, cwd=cwd, env=env, capture_output=True, text=True, timeout=timeout)
    except subprocess.TimeoutExpired:
        r = None
        block = f'==== {title} ====\nTIMEOUT после {timeout}s\n'
        LOG.write_text((LOG.read_text() if LOG.exists() else '') + block)
        results.append((title, 'TIMEOUT', timeout))
        print(f'{title}: TIMEOUT')
        return
    dt = time.time() - t0
    block = f'==== {title} ====\n$ {" ".join(cmd)}\n{r.stdout}\n{r.stderr}\n--- rc={r.returncode} ({dt:.0f}s)\n\n'
    LOG.write_text((LOG.read_text() if LOG.exists() else '') + block)
    results.append((title, 'rc=' + str(r.returncode), f'{dt:.0f}s'))
    print(f'{title}: rc={r.returncode}, {dt:.0f}s')

LOG.write_text(f'POTOK run {time.strftime("%Y-%m-%d %H:%M:%S")}\n\n')
run('корневой модуль: go test ./...', ['/usr/local/go/bin/go','test','./...'], cwd=RUN_DIR)
flat = RUN_DIR + '/flat'
if os.path.exists(flat + '/go.mod'):
    run('flat: go test -race ./...', ['/usr/local/go/bin/go','test','-race','./...'], cwd=flat)
    run('flat: Zero-GC (TestZero)', ['/usr/local/go/bin/go','test','-run','TestZero','./bus/...','./front/...'], cwd=flat)
    run('flat: сборка примеров', ['/usr/local/go/bin/go','build','-o','/dev/null','./examples/simple/...','./examples/drone/...'], cwd=flat)
    env0 = dict(env, CGO_ENABLED='0')
    run('flat: сборка bin/organism', ['/usr/local/go/bin/go','build','-ldflags','-s -w','-o','bin/organism','./cmd/organism/...'], cwd=flat, env=env0)
else:
    print('flat/ не найден — пропуск полного прогона (это скелетная ветка)')

print()
print('СВОДКА:')
for t, rc, dt in results:
    print(' -', t, '|', rc, '|', dt)
print('лог:', LOG)
```

```python
# 4. Копия лога на Диск (если он смонтирован ячейкой 3). Без монтирования лог умрёт вместе с сессией.
import shutil, os
src = '/content/potok-run.log'
dst = '/content/drive/MyDrive/ОБОЗ/colab/potok-run.log'
if os.path.isdir('/content/drive/MyDrive'):
    os.makedirs(os.path.dirname(dst), exist_ok=True)
    shutil.copy(src, dst)
    print('сохранено:', dst)
else:
    print('Диск не смонтирован. Лог останется в /content/potok-run.log до закрытия сессии.')
    print('Чтобы сохранить: выполните ячейку 3, затем эту ещё раз.')
```
