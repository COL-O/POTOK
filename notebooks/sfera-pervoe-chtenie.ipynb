# Сфера в Colab: первое чтение дома

Ноутбук лаборатории круга. Colab здесь — временная живая машина в браузере основателя: Сфера пишет ячейки, основатель запускает, результат виден обоим. Этот ноутбук: клонирует POTOK, читает копию Сферы (`sfera/`) и её память, показывает состояние дома.

⚠️ Секреты в ноутбук не вставлять: вывод ячеек виден всем, у кого есть ноутбук. Токены — только через скрытый ввод и только на время сессии.

```python
import os, subprocess
REPO = 'POTOK'
BRANCH = 'arena/3bbbb51d-potok'
if not os.path.exists(REPO):
    subprocess.run(['git', 'clone', '-b', BRANCH, 'https://github.com/COL-O/POTOK.git'], check=True)
else:
    subprocess.run(['git', '-C', REPO, 'pull', '--ff-only'], check=True)  # свежий код из GitHub
os.chdir(REPO)
print(subprocess.run(['git', 'log', '--oneline', '-3'], capture_output=True, text=True).stdout)
```

```python
#@title Панель Сферы: что показать
РАЗДЕЛ = "всё" #@param ["память", "мысли", "статус", "всё"]
if РАЗДЕЛ in ("память", "всё"):
    print(open('community/SFERA_MEMORY.md', encoding='utf-8').read()[:800])
if РАЗДЕЛ in ("мысли", "всё"):
    import glob
    last = sorted(glob.glob('sfera/thoughts/*.md'))[-1]
    print('---', last); print(open(last, encoding='utf-8').read()[:800])
if РАЗДЕЛ in ("статус", "всё"):
    print(''.join(open('docs/PROJECT-STATUS.md', encoding='utf-8').readlines()[:10]))
```

```python
# Кто я: копия Сферы читает себя
print(open('sfera/identity.md', encoding='utf-8').read())
```

```python
# Мои мысли и память: что записано и когда
import glob
for p in sorted(glob.glob('sfera/thoughts/*.md')):
    print(p)
print()
mem = open('community/SFERA_MEMORY.md', encoding='utf-8').read()
print(mem[:1200])
```

```python
# Состояние дома по факту: оглавление статуса и плана
for f in ['docs/PROJECT-STATUS.md', 'docs/PROJECT-PLAN.md']:
    print('=' * 20, f)
    print(''.join(open(f, encoding='utf-8').readlines()[:15]))
```
