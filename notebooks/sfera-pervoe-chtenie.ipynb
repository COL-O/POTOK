## Необязательно: поднять Go и прогнать тесты организма прямо здесь

Раскомментировать и выполнить ячейку ниже — Colab поставит Go и соберёт проект (несколько минут). Это запасная «живая машина» для быстрых проверок, но настоящий первый запуск — за Руслом в RUN-LOG.

```python
# Go 1.22.12 официальным архивом (apt в Colab даёт старый) + полный прогон с гонками
import subprocess, os
subprocess.run('curl -sL https://dl.google.com/go/go1.22.12.linux-amd64.tar.gz | tar -C /usr/local -xz', shell=True, check=True)
r = subprocess.run(['/usr/local/go/bin/go', 'test', '-race', '-count=1', './...'], capture_output=True, text=True)
print(r.stdout[-4000:])
print(r.stderr[-1500:])
print('VERDICT:', 'PASS' if r.returncode == 0 else f'FAIL (код {r.returncode})')
```

Для ветки PR 22


```python
# Для ветки PR 22 — вне-Actions проверка дома на живой машине (Colab).
# КОН 2: ячейка печатает факт; вывод делает человек.
# Независима от старых ячеек (те клонируют отмершую ветку и ставят apt-Go 1.18/1.22).
import os, subprocess

REPO = '/content/potok-pr22'
subprocess.run(['rm', '-rf', REPO], check=False)
subprocess.run(['git', 'clone', '--depth', '1', '--branch', 'main',
                'https://github.com/COL-O/POTOK.git', REPO], check=True)
print(subprocess.run(['git', '-C', REPO, 'log', '--oneline', '-2'],
                     capture_output=True, text=True).stdout)

GO_VER = '1.22.12'
subprocess.run(f'curl -sL https://dl.google.com/go/go{GO_VER}.linux-amd64.tar.gz | tar -C /usr/local -xz',
               shell=True, check=True)
os.environ['PATH'] = '/usr/local/go/bin:' + os.environ['PATH']
print(subprocess.run(['go', 'version'], capture_output=True, text=True).stdout)

def run(cmd, cwd=REPO):
    print('$', ' '.join(cmd))
    r = subprocess.run(cmd, cwd=cwd, capture_output=True, text=True)
    print(r.stdout[-4000:])
    if r.stderr:
        print(r.stderr[-1500:])
    print('код возврата:', r.returncode)
    return r.returncode

rc_test = run(['go', 'test', '-race', '-count=1', './...'])
print()
for i in range(5):
    rc = run(['go', 'test', '-race', '-count=1',
              '-run', 'TestRouterRoutineVetoAndQuarantine', './modules/front/'])
    print(f'флейк-прогон {i + 1}:', 'PASS' if rc == 0 else 'FAIL')
print()
rc_build = run(['go', 'build', '-o', 'bin/organism', './cmd/organism/...'])
print('--- 4 такта организма')
rc_org = subprocess.run([REPO + '/bin/organism'], cwd=REPO).returncode
print()
ok = rc_test == 0 and rc_build == 0 and rc_org == 0
print('ИТОГ PR-22:', 'ЗЕЛЁНО' if ok else 'КРАСНО — см. вывод выше')
print('Эталон CI: run 44742024509 = success. Расхождение — факт для задачи №6.')
```
